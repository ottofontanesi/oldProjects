```python
import pandas as pd
from pandas import ExcelWriter
from pathlib import Path
```

```python
input_path = Path("C:\\Users\\fontanesio\\Documents\\pythonScripts\\progetti\\personali\\LogPPT-master\\tmp\\test.xlxs")
writer = ExcelWriter(input_path)
def save_xls(df:pd.DataFrame, writer:ExcelWriter, key:str, close:bool=False) -> None:
    """
    Save a dictionary of dataframes to an excel file, 
    with each dataframe as a separate page
    :param df: pd.DataFrame or pd.series to export
    :param writer: excel writer
    :param key: sheet name
    """
    df.to_excel(writer, sheet_name=key)
    if close:
        writer.save()
    return None
```
