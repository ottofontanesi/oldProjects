```python
import os
from pathlib import Path
import textract
import pandas as pd
import re
```

```python
_SRC_PATH_ = Path("C:\\Users\\fontanesio\\Desktop\\otto\\miscellanee\\lettere")
anagr = pd.read_csv(_SRC_PATH_ / "anagrafica.txt", sep=";", dtype={"NomeCompleto":str, "Codice_dip":str, "Codice_Fiscale":str})
anagr.head()
```

```python
print("Numero files in {} : {}".format(_SRC_PATH_ / "lett_bkp", len(os.listdir(_SRC_PATH_ / "lett_bkp"))))

for pdf_file_name in os.listdir(_SRC_PATH_ / "lett_bkp"):
    try:
        print("Working on : {}".format(_SRC_PATH_ / "lett" / pdf_file_name))
        text = textract.process("C:\\Users\\fontanesio\\Desktop\\otto\\miscellanee\\lettere\\lett_bkp\\" + pdf_file_name).decode("utf-8")
        nome = re.search(r"Gentile\r\n(.*?)\r\n", text).group(1)
        codice_matr = anagr[anagr.NomeCompleto == nome].Codice_dip.values[0]
        codice_fisc = anagr[anagr.NomeCompleto == nome].Codice_Fiscale.values[0]
        raw_file_name = pdf_file_name.split(" ")[0]
        new_pdf_file_name = raw_file_name+codice_matr+"_"+codice_fisc+"_premio_2020.pdf"
        os.rename("C:\\Users\\fontanesio\\Desktop\\otto\\miscellanee\\lettere\\lett_bkp\\" + pdf_file_name, "C:\\Users\\fontanesio\\Desktop\\otto\\miscellanee\\lettere\\lett_bkp\\" + new_pdf_file_name)
    except:
        print(pdf_file_name)
```
