```python
def update_value(df_to_update, data):
    print("..updating..\nKEYS : IDX = {}, ATTR = {}\tOLD = {}\tNEW = {}".format(data.IDX, data.ATTR, df_to_update.loc[data.IDX, data.ATTR], data.VALORE))
    df_to_update.loc[data.IDX, data.ATTR] = data.VALORE
    return None

s2.apply(lambda x: update_value(df_to_update=s1, data=x), axis=1)
```
